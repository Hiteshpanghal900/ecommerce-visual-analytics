## Since Review score is unavailbale it would be better to drop it since replacing by mean or mode will not work

```python
df = df.dropna(subset=['review_score'])
```

```python
order_cols = [
    'order_id',
    'order_status',
    'total_item_value',
    'total_freight_value',
    'customer_state',
    'actual_delivery_days',
    'delivery_delay_days',
    'late_delivery_flag',
    'review_score',
    'review_risk_flag'
]

order_df=df[order_cols]
```

```python
order_df
```

#### Understanding Columns 
- delivery_delay_columns: here if delivery_delay_columns is negative then product delivered early if 0 means then product delivered at exact time if positive then product delivered late
- total_freight_value: means total cost for shipping
- customer_state: state of Brazi where customer lives
- late_delievery_flag: True if product delivered late False otherwise
-  review_risk_flag: True if review is less than particular threshold of review_score otherwise false
  

```python
order_df.info()
```

```python
print("Shape of table: ",order_df.shape)
order_df.isna().sum()
```

- columns like actual_delivery_days and delivery_delay_days cannot be confidently replaced because the item was never reached to customer so better to drop them
- 

```python
order_df = order_df[
    order_df["actual_delivery_days"].notna()
].copy()
```

```python
order_df.isna().sum()
```

Now our dataset is almost clean for columns: 'order_status', 'total_item_value', 'total_freight_value',
       'customer_state', 'actual_delivery_days', 'delivery_delay_days',
       'late_delivery_flag', 'review_score', 'review_risk_flag'

```python
df.isna().sum()
```

```python
new_df = df[df["order_id"].isin(order_df["order_id"])].copy()
```

```python
new_df.shape
```

```python
new_df.isna().sum()
```

```python
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_cols:
    new_df[col] = pd.to_datetime(new_df[col], errors="coerce")
```

```python
new_df["order_status"].value_counts()
```

```python
new_df["review_score"].value_counts().sort_index()
```

```python
new_df["late_delivery_flag"].value_counts()
```

```python
total_orders = new_df["order_id"].nunique()
avg_rating = new_df["review_score"].mean()
late_percentage = new_df["late_delivery_flag"].mean() * 100
avg_order_value = new_df["total_item_value"].mean()
avg_freight = new_df["total_freight_value"].mean()
```

```python
print("Total Orders:", total_orders)
print("Average Rating:", avg_rating)
print("Late Delivery %:", late_percentage)
print("Average Order Value:", avg_order_value)
print("Average Freight:", avg_freight)
```
